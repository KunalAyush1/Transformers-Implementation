<h3>Data Cleaning</h3>

```python


text = text.lower() #lowercasing all of them

text = re.sub(r"\[\d+\]","",text) #removing references like [1],etc
text = re.sub(r"[^a-z\s]","",text) #removing punctuations and numbers

tokens = text.split()

print(tokens[:50])
print(f"Total tokens are {len(tokens)}")
print(f"Vocabulary size is {len(set(tokens))}")

```

```python
stopwords = {
    "the","is","are","a","an""of", "and", "to", "in", "as",
    "for", "with", "on", "by", "that", "it", "or", "from", "also",
    "have", "has", "be", "been", "being"
}
tokens = [word for word in tokens if word not in stopwords]
```

```python
tokens
```

```python
#creating a vocabulary

vocab = sorted(set(tokens))

word_to_idx = {word: i for i, word in enumerate(vocab)}
idx_to_word = {i: word for word, i in word_to_idx.items()}

vocab_size = len(vocab)

print(f"Total tokens: {len(tokens)}")
print(f"Total vocab size: {vocab_size}")
```

```python
encoded = [word_to_idx[word] for word in tokens]
```

```python
#making traning data
window_size = 2
data = []

for i in range(window_size, len(encoded) - window_size):
    context = []
    
    for j in range( i - window_size, i + window_size + 1):
        if j != i:
            context.append(encoded[j])
    target = encoded[i]
    data.append((context, target))
    
print("Training examples: ", len(data))

for context, target in data[:5]:
    context_words = [idx_to_word[i] for i in context]
    target_word = idx_to_word[target]
    print(f"{context_words} -> {target_word}")
```

```python
def one_hot(indices, vocab_size):
    vectors = torch.zeros(len(indices), vocab_size)
    
    for row, idx in enumerate(indices):
        vectors[row][idx] = 1.0
        
    return vectors
```

```python
#example

context = data[0][0]
x = one_hot(context, vocab_size)

print(x.shape)
```

<h2>Model</h2>

```python
class CBOW(nn.Module):
    def __init__(self, vocab_size, embedding_dim):
        super().__init__()
        
        self.W1 = nn.Parameter(torch.randn(vocab_size, embedding_dim) * 0.01)
        self.W2 = nn.Parameter(torch.randn(embedding_dim,vocab_size) * 0.01)
        
    def forward(self, x):
        # shape of x  is [context_size, vocab_size]
        
        embeddings = x @ self.W1
        
        hidden = torch.mean(embeddings, dim=0)
        #shape is (embedding_dim)
        
        output = hidden @ self.W2
        #shape is vocab_size
        
        return output
```

```python
embedding_dim = 50

model = CBOW(vocab_size, embedding_dim)

EPOCHS = 800

loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

loss_history = []


for epoch in range(EPOCHS):
    total_loss = 0
    
    for context, target in data:
        x = one_hot(context, vocab_size)
        y = torch.tensor([target], dtype=torch.long)
        
        output = model(x)
        
        output = output.unsqueeze(0)
        
        loss = loss_fn(output, y)
        
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        total_loss += loss.item()
    
    loss_history.append(total_loss)
        
    if epoch % 50 == 0:
        print(f"Epoch {epoch}, loss: {total_loss:.4f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.plot(loss_history)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("CBOW Training loss")
plt.show()
```

<h1>Visualising the embeddings using PCA</h1>

```python
from sklearn.decomposition import PCA
from mpl_toolkits.mplot3d import Axes3D
import numpy as np


embeddings = model.W1.data

selected_words = [
    "ai", "agents", "tools", "models", "language",
    "memory", "planning", "reason", "context",
    "systems"
]

selected_words = [word for word in selected_words if word in word_to_idx]


selected_vectors = torch.stack(
    [embeddings[word_to_idx[word]]
     for word in selected_words
     ]
)

pca = PCA(n_components = 3)
reduced = pca.fit_transform(selected_vectors.numpy())


```

```python
fig = plt.figure(figsize = (10,8))
ax = fig.add_subplot(111, projection='3d')

for i, word in enumerate(selected_words):
    x, y, z = reduced[i]
    
    ax.scatter(x, y, z)
    ax.text(x, y, z, word)
    
ax.set_title("3D PCA Visualization of CBOW Word Embeddings")
ax.set_xlabel("PC1")
ax.set_ylabel("PC2")
ax.set_zlabel("PC3")

plt.show()
```
